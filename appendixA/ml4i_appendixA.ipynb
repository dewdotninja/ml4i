# A ติดตั้งและใช้งานเครื่องมือซอฟต์แวร์

โค้ดทั้งหมดในหนังสือนี้เขียนโดยใช้ภาษาไพทอน และใช้เครื่องมือหลักดังนี้

| เครื่องมือ | หน้าที่ | บทที่ใช้ |
|---|---|---|
| **NumPy** | การคำนวณเชิงเลข แอเรย์และเมทริกซ์ | ทุกบท |
| **matplotlib** | การพล็อตกราฟและสร้างภาพนามธรรม | ทุกบท |
| **pandas** | การนำเข้าและใช้งานข้อมูล | บทที่ 1 - 7 |
| **scikit-learn** | โมเดลเชิงคลาสิก การประมวลผลเบื้องต้น (preprocessing) สายท่อ (pipeline) การประเมิน (evaluation) | บทที่ 1 - 7 |
| **PyTorch** | เทนเซอร์ การหาอนุพันธ์อัตโนมัติ โครงข่ายประสาทเทียม  | บทที่ 8 - 16 |
| **SciPy** | การประมวลสัญญาณและสถิติ | ส่วนที่เกี่ยวข้อง |

โดยอาจมีการใช้แพ็กเกจอื่นในบางส่วน ซึ่งจะระบุเพิ่มเติมในเนื้อหาส่วนนั้น

ภาพรวมคือจะใช้ scikit-learn สำหรับข้อมูลเชิงตาราง (tabular data) หนึ่งแถวคือหนึ่งตัวอย่าง ส่วนคอลัมน์แทนลักษณะเด่นที่ผ่านการจัดการทางวิศวกรรม (engineered features) เรียกว่า ML เชิงคลาสิก แต่เมื่อข้อมูลมีจำนวนมิติสูง และเป็นข้อมูลดิบ หรือต้องการสถาปัตยกรรมของโมเดลเฉพาะ การใช้ PyTorch จะเหมาะสมกว่า 

## A.1 การตั้งค่าสภาวะแวดล้อม

สภาวะแวดล้อม (environment) ในทางคอมพิวเตอร์หมายถึงพื้นที่การทำงานที่มีการติดตั้งเครื่องมือซอฟต์แวร์ที่ใช้ทั้งหมดพร้อมสำหรับการประมวลผล โดยเราสามารถมีสภาวะแวดล้อมหลายตัวในเครื่องคอมพิวเตอร์เดียวกัน สำหรับรันงานที่ต้องการเครื่องมือ (หรือเวอร์ชันของเครื่องมือ) ที่แตกต่างกัน โดยก่อนจะเริ่มรันโค้ดจะต้องแอกทิเวต (activate) สภาพแวดล้อมเสียก่อน หรือกรณีที่ใช้โปรแกรม Visual Studio Code (VSC) ต้องเลือกเคอร์เนลที่สร้างสำหรับงานนั้น

ตัวอย่างเช่นต้องการสร้างสภาวะแวดล้อมชื่อ ml4i สำหรับรันโน๊ตบุกในหนังสือ ในกรณีที่ใช้ conda ต้องติดตั้ง anaconda หรือ miniconda ก่อน แล้วใช้คำสั่งใน terminal ดังนี้

```console
conda create -n ml4i python=3.13 -y
conda activate ml4i
pip install numpy pandas scipy matplotlib scikit-learn torch jupyter
```

หรือกรณีที่ใช้เครื่องมือใหม่กว่าเช่น <a href="https://docs.astral.sh/uv/">uv</a>

```console
uv init ml4i && cd ml4i
uv add numpy pandas scipy matplotlib scikit-learn torch jupyter
uv run jupyter lab
```
